A. Word Rotation Problem - Moves the first letter of a string to the end.

Requirement: Use string indexing or slicing to construct the returned string.

```python
def rotate_word(text): #defines rotate_word() as a function.
    return text[1:] + text[0] #returns the second element up until the last element and then adds the first element.
```

```python
rotate_word("pyhton")
```

```python
rotate_word("logic")
```

```python
rotate_word("Code")
```

```python
rotate_word("A")
```

B. Username Builder Problem - Create a username that converts two strings into lowercases, removing the space, and concatenating it with a "." .

Requirement: Use basic string methods and string concatenation. Return the completed username

```python
def make_username(first_name, last_name): #defines make_username() as a function that accepts two strings that are considered first_name and last_name
    first_name = first_name.lower() #converts the letters of the first_name to lowercase
    first_name = first_name.replace(" ", "") #removes any spaces in first_name
    
    last_name = last_name.lower() #converts the letters of the last_name to lowercase
    last_name = last_name.replace(" ", "") #removes any spaces in last_name
    
    return first_name + "." + last_name #returns the updated first_name, concatenates it with a ".", and then the updated last_name.
```

```python
make_username("Ada", "Lovelace")
```

```python
make_username("Alan", "Turing")
```

```python
make_username("Ana Maria", "De Leon") 
```
